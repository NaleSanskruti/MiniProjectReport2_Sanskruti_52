```python
# Mini Project 2

# Student Database Management System Using Hashing

class StudentDatabase:
    def __init__(self, size=10):
        self.size = size
        self.hash_table = [[] for _ in range(size)]

    # Hash function
    def hash_function(self, student_id):
        return student_id % self.size

    # Insert student record
    def insert_student(self, student_id, name, course, marks):
        index = self.hash_function(student_id)

        # Check if student already exists
        for student in self.hash_table[index]:
            if student["id"] == student_id:
                print("Student ID already exists!")
                return

        student = {
            "id": student_id,
            "name": name,
            "course": course,
            "marks": marks
        }

        self.hash_table[index].append(student)
        print("Student record inserted successfully!")

    # Search student record
    def search_student(self, student_id):
        index = self.hash_function(student_id)

        for student in self.hash_table[index]:
            if student["id"] == student_id:
                print("\nStudent Found!")
                print("Student ID:", student["id"])
                print("Name:", student["name"])
                print("Course:", student["course"])
                print("Marks:", student["marks"])
                return

        print("Student record not found!")

    # Display all student records
    def display_students(self):
        print("\n--- Student Database ---")

        found = False

        for index, bucket in enumerate(self.hash_table):
            for student in bucket:
                found = True
                print(
                    "ID:", student["id"],
                    "| Name:", student["name"],
                    "| Course:", student["course"],
                    "| Marks:", student["marks"]
                )

        if not found:
            print("Database is empty.")

    # Update student record
    def update_student(self, student_id):
        index = self.hash_function(student_id)

        for student in self.hash_table[index]:
            if student["id"] == student_id:
                student["name"] = input("Enter new name: ")
                student["course"] = input("Enter new course: ")
                student["marks"] = float(
                    input("Enter new marks: ")
                )

                print("Student record updated successfully!")
                return

        print("Student record not found!")

    # Delete student record
    def delete_student(self, student_id):
        index = self.hash_function(student_id)

        for student in self.hash_table[index]:
            if student["id"] == student_id:
                self.hash_table[index].remove(student)
                print("Student record deleted successfully!")
                return

        print("Student record not found!")


# Main program
db = StudentDatabase()

while True:
    print("\n===== Student Database Management =====")
    print("1. Insert Student")
    print("2. Search Student")
    print("3. Display All Students")
    print("4. Update Student")
    print("5. Delete Student")
    print("6. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        try:
            student_id = int(input("Enter Student ID: "))
            name = input("Enter Student Name: ")
            course = input("Enter Course: ")
            marks = float(input("Enter Marks: "))

            if not 0 <= marks <= 100:
                print("Marks must be between 0 and 100.")
                continue

            db.insert_student(
                student_id, name, course, marks
            )
        except ValueError:
            print("Please enter valid numeric values.")

    elif choice == "2":
        try:
            student_id = int(input("Enter Student ID to search: "))
            db.search_student(student_id)
        except ValueError:
            print("Please enter a valid Student ID.")

    elif choice == "3":
        db.display_students()

    elif choice == "4":
        try:
            student_id = int(input("Enter Student ID to update: "))
            db.update_student(student_id)
        except ValueError:
            print("Please enter valid numeric values.")

    elif choice == "5":
        try:
            student_id = int(input("Enter Student ID to delete: "))
            db.delete_student(student_id)
        except ValueError:
            print("Please enter a valid Student ID.")

    elif choice == "6":
        print("Thank you for using Student Database!")
        break

    else:
        print("Invalid choice! Please try again.")
```
